```python
import numpy as np
import random
```

```python
# 0 = path, -1 = wall, 100 = goal

maze = [
    [0, 0, 0, -1],
    [0, -1, 0, -1],
    [0, 0, 0, 100]
]

rows = len(maze)
cols = len(maze[0])

start = (0, 0)
goal = (2, 3)
```

```python
def state_to_index(state):
    return state[0] * cols + state[1]

def index_to_state(index):
    return (index // cols, index % cols)

num_states = rows * cols
num_actions = 4  # up, down, left, right
```

```python
Q = np.zeros((num_states, num_actions))
gamma = 0.9
alpha = 0.8
epsilon = 0.2
```

```python
actions = [(-1,0), (1,0), (0,-1), (0,1)]  # up, down, left, right

def is_valid(state):
    r, c = state
    return 0 <= r < rows and 0 <= c < cols and maze[r][c] != -1
```

```python
def get_reward(state):
    r, c = state
    if maze[r][c] == 100:
        return 100
    return -1
```

```python
def choose_action(state_index):
    if random.uniform(0,1) < epsilon:
        return random.randint(0,3)
    return np.argmax(Q[state_index])
```

```python
def move(state, action):
    dr, dc = actions[action]
    next_state = (state[0] + dr, state[1] + dc)

    if is_valid(next_state):
        return next_state
    return state
```

```python
for episode in range(1000):
    state = start

    while state != goal:
        state_index = state_to_index(state)
        action = choose_action(state_index)

        next_state = move(state, action)
        reward = get_reward(next_state)

        next_index = state_to_index(next_state)

        Q[state_index][action] = reward + gamma * np.max(Q[next_index])

        state = next_state
```

```python
state = start
path = [state]

while state != goal:
    state_index = state_to_index(state)
    action = np.argmax(Q[state_index])
    state = move(state, action)
    path.append(state)

print("Optimal Path:", path)‹
```
