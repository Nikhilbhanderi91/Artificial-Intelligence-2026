```python
import nltk
from nltk.util import ngrams
from collections import Counter

nltk.download('punkt')
```

```python
text = """Natural language processing is very useful.
It helps computers understand human language.
Natural language processing is widely used."""
```

```python
nltk.download('punkt_tab', quiet=True)
tokens = nltk.word_tokenize(text.lower())
print("Tokens:\n", tokens)
```

```python
def generate_ngrams(tokens, n):
    return list(ngrams(tokens, n))
```

```python
unigrams = generate_ngrams(tokens, 1)
print("\nUnigrams:\n", unigrams)
```

```python
bigrams = generate_ngrams(tokens, 2)
print("\nBigrams:\n", bigrams)
```

```python
trigrams = generate_ngrams(tokens, 3)
print("\nTrigrams:\n", trigrams)
```

```python
bigram_freq = Counter(bigrams)
print("\nBigram Frequencies:\n", bigram_freq)
```

```python
def predict_next_word(seed_word, bigrams):
    candidates = [word2 for (word1, word2) in bigrams if word1 == seed_word]
    if candidates:
        return Counter(candidates).most_common(1)[0][0]
    else:
        return "No prediction"

# Example Prediction
seed = "natural"
print("\nNext word after '{}':".format(seed), predict_next_word(seed, bigrams))
```

```python
import nltk
from nltk.util import ngrams
from collections import Counter
```

```python
text = "The cat sat on the mat. The cat is very cute. The cat likes milk."
```

```python
tokens = nltk.word_tokenize(text.lower())
bigrams = list(ngrams(tokens, 2))
```

```python
bigram_freq = Counter(bigrams)
```

```python
def predict_next(seed):
    candidates = [(w2, count) for (w1, w2), count in bigram_freq.items() if w1 == seed]
    if candidates:
        return sorted(candidates, key=lambda x: x[1], reverse=True)[0][0]
    return "No prediction"

print("Next word after 'the':", predict_next("the"))
```
